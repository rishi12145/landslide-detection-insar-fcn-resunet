# Landslide Detection using InSAR Data with FCN and ResUNet Models

## Detailed Comparison

This notebook presents a detailed comparison between two deep learning models:
- Fully Convolutional Network (FCN)
- ResUNet

Both models are trained on multi-temporal InSAR data to detect landslide-prone regions.

The comparison is based on:
- Accuracy
- Precision
- Recall
- F1 Score
- Dice Score
- IoU (Intersection over Union)

The goal is to analyze which model performs better for this specific dataset and understand the reasons behind the performance differences.

```python
import matplotlib.pyplot as plt
import numpy as np

# ================= VALUES =================

metrics = ["Accuracy", "Precision", "Recall", "F1", "Dice", "IoU"]

# FCN
fcn = [0.9908, 0.8928, 0.8641, 0.8736, 0.8737, 0.7834]

# ResUNet
res = [0.9839, 0.8234, 0.7676, 0.7881, 0.7881, 0.6583]

x = np.arange(len(metrics))

plt.figure(figsize=(10,5))

plt.bar(x - 0.2, fcn, 0.4, label="FCN")
plt.bar(x + 0.2, res, 0.4, label="ResUNet")

plt.xticks(x, metrics, rotation=30)
plt.ylabel("Score")
plt.title("Detailed Model Comparison")
plt.legend()
plt.grid()

plt.show()
```

```python
diff = [f - r for f, r in zip(fcn, res)]

plt.figure(figsize=(8,5))
plt.bar(metrics, diff)

plt.axhline(0)
plt.title("Performance Difference (FCN - ResUNet)")
plt.ylabel("Difference")

plt.grid()
plt.show()
```

## FCN vs ResUNet: Performance Analysis

From the results, FCN performs better than ResUNet across all major metrics such as Dice, IoU, and F1 Score.

This can be explained based on the following observations:

### 1. Simpler Architecture Advantage
FCN has a relatively simpler architecture compared to ResUNet.  
Because of this, it is able to learn the dataset patterns more easily without over-complicating the feature extraction process.

In contrast, ResUNet is deeper and more complex, which may require more data and fine-tuning to perform at its best.


### 2. Better Generalization on Dataset
The dataset used in this project is moderate in size.  
FCN, being simpler, generalizes better on such datasets.

ResUNet, on the other hand, may:
- Overfit easily  
- Or fail to fully utilize its depth  


### 3. Training Stability
FCN training was more stable:
- Smooth loss convergence  
- Consistent accuracy improvement  

ResUNet showed:
- Slower learning initially  
- More sensitivity to training parameters  


### 4. Feature Learning vs Complexity
While ResUNet is designed to capture deeper features using residual connections,  
in this case, the added complexity did not translate into better performance.

This suggests that:
- The problem does not require very deep feature extraction  
- A well-structured simpler model (FCN) is sufficient

  
### 5. Data Compatibility
The FCN model aligns well with the input representation:
- Coherence + cos(phase) + sin(phase  

This combination works effectively with convolution-based architectures like FCN.


### Final Insight

FCN outperformed ResUNet not because it is more powerful,  
but because it is **more suitable for this specific dataset and task**.
